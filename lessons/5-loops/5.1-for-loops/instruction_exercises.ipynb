### for loop

Code that repeats an indented block once for each value supplied to it.

The `for` begins the loop.

Look at `for` on line 1.

```python
for number in range(3):
    print(number)
```

### range

`range()` supplies whole numbers. With one argument, it starts at zero and stops before that number.

The call supplies `0`, `1`, then `2`.

Look at `range(3)` on line 1.

```python
for number in range(3):
    print(number)
```

### Numbers from range

`range(3)` supplies three numbers: `0`, `1`, and `2`. The stopping number is excluded.

```python
for number in range(3):
    print(number)
```

Each time through the loop, `number` receives the next value. The colon begins an indented block. `range(3)` supplies numbers without making a list. Its stopping number, `3`, is excluded.

A `for` loop receives successive values and finishes when there are no more.

### loop variable

A name that receives the next value each time through the loop.

The `person` receives each item from `names`.

Look at `person` on line 2.

```python
names = ["Alex", "Sam"]
for person in names:
    print(person)
```

### loop body

The indented code that runs for each value in a loop.

The line is inside the loop. The final print is outside it.

Look at `print(person)` on line 3.

```python
names = ["Alex", "Sam"]
for person in names:
    print(person)
print("Done")
```

### One name at a time

The list supplies its items in order. The loop variable receives an item, not its position.

```python
names = ["Alex", "Jordan", "Sam"]
for name in names:
    print(name)
print("Everyone is greeted")
```

The first pass uses `"Alex"`, then `"Jordan"`, then `"Sam"`. Direct list iteration does not call `range()` or look up numeric positions. Both examples use the same `for` pattern: receive the next value, then run the body. A line aligned with `for` is outside the loop.

### Decide for each score

The `if` block makes a fresh decision for each score. Both indentation levels matter.

```python
scores = [55, 80, 65]
for score in scores:
    if score >= 60:
        print(score)
```

The loop visits every score. The condition controls whether that score is printed. A score that fails the condition does not stop the loop.

### accumulator

A variable that keeps a growing result as a loop runs, such as a total or a count.

The line adds this item to the running total.

Look at `total = total + points` on line 3.

```python
total = 0
for points in [2, 3]:
    total = total + points
print(total)
```

### Count matching scores

Start the count before the loop. Increase it only when the condition is true.

```python
scores = [55, 80, 65]
count = 0
for score in scores:
    if score >= 60:
        count = count + 1
print(count)
```

An accumulator can hold a count or a total. A count adds one for each match. A total adds the item itself. Reset it before the loop so rerunning the cell starts fresh.

### Collect matching scores

Start with an empty list. `.append()` adds one matching score to the new list.

```python
scores = [55, 80, 65]
passing = []
for score in scores:
    if score >= 60:
        passing.append(score)
print(passing)
```

Visit the original list while building a separate result. Keep initialization outside the loop. Each rerun should build a fresh result, not add to the previous run.

### Return the completed list

Put the same steps in a function. `return` belongs after the loop so every score is considered.

```python
def passing_scores(scores):
    passing = []
    for score in scores:
        if score >= 60:
            passing.append(score)
    return passing

print(passing_scores([55, 80, 65]))
```

`passing` is created anew on each function call. The function returns a new list and leaves its input unchanged. With an empty input, the body runs zero times and the function returns the empty result.

### nested loop

A loop inside another loop. The inside loop runs through its values for each outside value.

The inside loop runs again for each fruit.

Look at `for color in ["red", "green"]:` on line 2.

```python
for fruit in ["apple", "pear"]:
    for color in ["red", "green"]:
        print(color, fruit)
```

### Optional reference: positions and items

Use positions when the task needs them. `len(names)` gives the number of items, so these positions are valid.

```python
names = ["Alex", "Sam"]
for index in range(len(names)):
    print(index, names[index])
```

The direct version uses `for name in names:` when you only need the items. The indexed version supplies numbers and looks up each item. They are different ways to write a loop, not different descriptions of what happens inside Python.

### Optional reference: changing the step

The third argument controls the change between numbers. Here, add two each time and stop before seven.

```python
for number in range(1, 7, 2):
    print(number)
```

This produces `1`, `3`, `5`. With a negative step, the numbers go downward. For example, `range(3, 0, -1)` supplies `3`, `2`, `1`. Use the optional countdown task to try it.

## Terminology reference

**for loop:** Code that repeats an indented block once for each value supplied to it.

**range:** `range()` supplies whole numbers. With one argument, it starts at zero and stops before that number.

**loop variable:** A name that receives the next value each time through the loop.

**loop body:** The indented code that runs for each value in a loop.

**accumulator:** A variable that keeps a growing result as a loop runs, such as a total or a count.

**nested loop:** A loop inside another loop. The inside loop runs through its values for each outside value.
