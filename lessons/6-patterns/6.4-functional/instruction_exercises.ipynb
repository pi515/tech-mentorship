### Return text instead of printing it

````python
def make_message(name):
    return "Hello " + name

print(make_message("Jenna"))
print(make_message("Jenna"))
````

Purity applies to the function being discussed, not to the entire program that calls it.

### impure function

A function that does more than return an answer, or uses outside information that can change.

`append()` changes the original order. Previewing twice adds two bags.

Look at `items.append("bag")` on line 2.

````python
def preview_order(items):
    items.append("bag")
    return items
order = ["book"]
print(preview_order(order))
print(preview_order(order))
````

### A preview that changes the order

A preview should show an order without adding items. This function changes the original list each time you look.

- First preview: one bag is added.
- Second preview: another bag is added.
- A later checkout now sees both bags in the order.

Should viewing an order change what you buy?

Run the example above from the beginning to start with a fresh order. `items` refers to the same list supplied as `order`. `append()` adds a value to that list. Returning the list does not undo the change. Even if you ignore the returned value, the order has already changed.

The surprising behavior is that simply asking to see the order changes what a later part of the program receives. Calling the preview twice adds two bags. Printing was a visible side effect in the earlier example. Here the lasting change to the supplied list is the problem.

A safer preview could return a new list and leave the order alone. Adding a bag to the actual order should be a separate, intentional action. The optional “preserve the supplied list” problem lets you practice that rewrite. Side effects are useful when intended, but a function's purpose should make them clear.

### idempotence

Doing something again after it is done makes no further change.

Clean the text, then clean that already-cleaned text. It stays the same.

Look at `clean(" Hi ")` on line 3, `clean(cleaned)` on line 5.

````python
def clean(text):
    return text.strip().lower()
cleaned = clean(" Hi ")
print(cleaned)
print(clean(cleaned))
````

### Apply the operation again

Compare one application with applying the result again. Which transformation stops changing?

````python
def normalize(text):
    return text.strip().lower()
def double(number):
    return number * 2

print(normalize("  HELLO  "))
print(normalize(normalize("  HELLO  ")))
print(double(3))
print(double(double(3)))
````

`strip()` removes whitespace from both ends, and `lower()` makes letters lowercase. Normalizing this result again changes nothing. For string inputs, normalization is idempotent. Doubling is pure and deterministic, but not idempotent in general. Calling `double(3)` twice with the original input is a different experiment from calling `double(double(3))`. One successful example cannot prove a property for every input.

### closure

An inner function that remembers values from the function around it.

`multiply` remembers `factor`, even after `make_multiplier` finishes.

Look at `factor` on line 1, `factor` on line 3, `multiply` on line 4.

````python
def make_multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply
times_three = make_multiplier(3)
print(times_three(4))
````

### Return a function that remembers

Each call to `make_multiplier` creates a function with its own enclosing `factor`.

````python
def make_multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)
print(double(5))
print(triple(5))
````

The inner function uses its parameter `number` and the enclosing name `factor`. Returning `multiply` returns the function without calling it. After `make_multiplier` finishes, the returned function retains access to its enclosing factor. A closure need not change state: these numeric multipliers behave as pure transformations. The factory is also higher-order because it returns a function.

### stateful function

A function that remembers information and updates it between calls.

This counter remembers its count and adds one each time.

Look at `count[0] = count[0] + 1` on line 4.

````python
def make_counter():
    count = [0]
    def next_count():
        count[0] = count[0] + 1
        return count[0]
    return next_count
next_count = make_counter()
print(next_count(), next_count())
````

### Independent counters remember state

Predict the three results. Each factory call creates a separate one-element list.

````python
def make_counter():
    count = [0]
    def next_count():
        count[0] = count[0] + 1
        return count[0]
    return next_count

first = make_counter()
second = make_counter()
print(first(), first(), second())
````

`count[0]` is the first value in the enclosing list. Each call changes that stored number. The counter is stateful and impure: calls with no arguments can produce different results and modify remembered state. The two counters have separate lists. The multiplier closure remembers a factor without changing it, so closures are not automatically impure.

### Optional reference: lambda notation

A lambda writes a small function as one expression. Prefer a descriptive named function in today's work.


## Terminology reference

**function:** A named set of steps you can run whenever you need them.

**function call:** Running a function by writing its name followed by parentheses.

**parameter:** A name a function uses for an input you give it.

**argument:** The value you give a function when you call it.

**return value:** The answer a function gives back to the code that called it.

**scope:** The parts of your code where you can use a particular variable.

**side effect:** Something a function does besides giving back an answer.

**pure function:** A function that gives the same answer for the same inputs and does nothing else outside itself.

**determinism:** Getting the same answer each time you give the same inputs.

**list:** A group of values kept in order inside square brackets.

**impure function:** A function that does more than return an answer, or uses outside information that can change.

**idempotence:** Doing something again after it is done makes no further change.

**higher-order function:** A function that takes another function as input or gives a function back.

**closure:** An inner function that remembers values from the function around it.

**stateful function:** A function that remembers information and updates it between calls.
