## Terminology

**string:** An ordered sequence of characters representing text.

**character:** One element of a string, including a letter, digit, punctuation mark, or space.

**string literal:** Text written directly in source code inside quotation marks.

**index:** An integer position used to select a character. The first index is `0`.

**length:** The number of characters in a string, including spaces.

**membership:** Whether text occurs within another string. `in` returns `True` or `False`.

**keyword:** A reserved word in Python syntax, such as `in`. It cannot be a variable name.

**substring:** A contiguous sequence of characters within a string.

**concatenation:** Combining strings into a new string using `+`.

**slice:** A selection using `[start:stop]`. Start is included and stop is excluded.

**expression:** Code that evaluates to a value, such as `len(firstName)` or `something[0:4]`.

**method:** A function accessed through a value using a dot, such as `firstName.upper()`.

**chaining:** Applying another method to the result of a method call.

**immutable:** Cannot be changed in place. String methods return new strings.

**negative index:** A position counted from the end of a string. `-1` selects its last character.

**delimiter:** Text used to separate pieces when splitting or joining a string.

**list:** An ordered collection of values. `split` returns a list of string pieces.

### Indexing selects one character

An **index** is a character's position. Python starts at **0**. Square brackets select the character. The gray code line numbers are separate from character indices.

````python
word = "Something"
#       012345678
print(word[0])
print(word[3])
````

The plural of index is **indices**. An address helps locate a house. An index helps locate a character. Unlike typical street numbers, string indices start at 0. Spaces count as characters too.

````python
print(len("MyString"))
print("s" in "string")
print("z" in "string")
````

### Slicing selects a substring

A **slice** uses `string[start:stop]`. Include the start. Exclude the stop.

````python
something = "Something"
#            012345678
print(something[0:4])
print(something[4:9])
````

`something[0:4]` selects indices 0, 1, 2, and 3: `"Some"`. It excludes index 4, the `"t"`. The second slice returns `"thing"`. Compare this with the four separate index lookups you just combined. `something[0:4]` is an **expression**: code that evaluates to a value.

### Practice reference: omitted bounds and negative indices

Omit a bound to use that end of the string. Negative indices count from the end. `-1` is the last character.

````python
something = "Something"
print(something[:4])
print(something[4:])
print(something[-5:])
````

For this nine-character string, `-9` is the first position and `-5` is the `"t"`. The stop is still excluded. `something[-9:-5]` is the same selection as `something[0:4]`. An expression such as `len(something) - 5` can be used as a bound too. With the default forward direction, `something[-1:-4]` is empty. Negative bounds do not automatically reverse a string.

### Practice reference: split, join, and delimiter

A **delimiter** separates pieces. `split` returns a list of strings. `join` combines pieces with the delimiter between them.

````python
text = "Make like a banana and split!"
pieces = text.split("banana")
print(pieces)
print("banana".join(pieces))
````

The delimiter here is the substring `"banana"`. The two pieces retain their spaces. The original text is unchanged. In `"".join(pieces)`, an empty delimiter adds nothing between pieces. You only need to read the returned list here. Lists receive their own lesson.

### Practice reference: replace

`replace(search, replacement)` returns text with matching substrings replaced. Search text is case-sensitive.

````python
text = "Make like a banana and split!"
print(text.replace("banana", "pear"))
print(text)
````

The new sentence contains `"pear"`. The original still contains `"banana"`. Use this method in the original `"Not Ice"` and `"N O I C E"` challenges. After practice, explain a slice to a mentor and save your notebook. You do not need to finish every challenge today.
